## CSV başlık kontrolü

İlk incelemede `Discount` ve `DLC count` sütunlarının başlıkta birleştiği görüldü.
Bu nedenle veri satırları yanlış sütunlarla eşleşiyordu. Birleşmiş başlık iki ayrı
sütuna ayrıldıktan sonra örnek veri 40 sütunla doğru şekilde okundu.

```python
kullanilacak_sutunlar = [
    "AppID",
    "Name",
    "About the game",
    "Price",
    "Windows",
    "Mac",
    "Linux",
    "Supported languages",
    "Categories",
    "Genres"
]

satir_sayisi = 0
eksik_degerler = pd.Series(0, index=kullanilacak_sutunlar)

for parca in pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kullanilacak_sutunlar,
    chunksize=10000
):
    satir_sayisi += len(parca)
    eksik_degerler += parca.isna().sum()

print("Toplam satır sayısı:", satir_sayisi)
eksik_degerler
```

```python
eksik_ozeti = pd.DataFrame({
    "Eksik değer": eksik_degerler,
    "Eksik oranı (%)": (eksik_degerler / satir_sayisi * 100).round(2)
})

eksik_ozeti
```

## İlk değerlendirme

Veri setinde toplam 125.855 oyun kaydı bulunmaktadır. Fiyat, işletim sistemi ve
desteklenen dil alanlarında eksik değer görülmemiştir. Oyun açıklaması, kategori
ve tür alanlarında ise eksik kayıtlar bulunmaktadır. Bu alanlar anlamsal aramada
kullanılacağı için veri temizleme aşamasında ayrıca değerlendirilecektir.

```python
eksik_oyun_listesi = []

for parca in pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kullanilacak_sutunlar,
    chunksize=10000
):
    eksik_mi = parca[
        ["About the game", "Categories", "Genres"]
    ].isna().any(axis=1)

    eksik_kayitlar = parca.loc[
        eksik_mi,
        ["AppID", "Name", "About the game", "Categories", "Genres"]
    ].copy()

    eksik_oyun_listesi.append(eksik_kayitlar)

eksik_oyunlar = pd.concat(eksik_oyun_listesi, ignore_index=True)

print("En az bir alanı eksik oyun sayısı:", len(eksik_oyunlar))
eksik_oyunlar.head(20)
```

```python
playtest_kayitlari = eksik_oyunlar[
    eksik_oyunlar["Name"]
    .fillna("")
    .str.contains("Playtest", case=False)
]

print("Eksik alanı bulunan Playtest sayısı:", len(playtest_kayitlari))
playtest_kayitlari[["AppID", "Name"]].head(20)
```

## Eksik kayıtların incelenmesi

Açıklama, kategori veya tür alanlarından en az biri eksik olan 9.568 kayıt
bulunmaktadır. Bu kayıtların 8.133 tanesinin adında `Playtest` ifadesi yer
almaktadır. Eksik kayıtların büyük bölümünün test sürümlerinden oluştuğu
görülmüştür.

Eksik olan bütün kayıtlar doğrudan silinmeyecektir. Playtest kayıtları ile normal
oyunlar veri temizleme aşamasında ayrı ayrı değerlendirilecektir.

```python
kontrol_sutunlari = [
    "Name",
    "About the game",
    "Categories",
    "Genres"
]

oyun_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kontrol_sutunlari
)

playtestler = oyun_kontrol[
    oyun_kontrol["Name"].str.contains(
        "Playtest", 
        case=False, 
        na=False 
    )
]

normal_oyunlar = oyun_kontrol[
    oyun_kontrol["Name"].str.contains(
        "Playtest",
        case=False,
        na=False
    ) == False
]

kullanilabilir_oyunlar = normal_oyunlar.dropna(
    subset=[
        "Name",
        "About the game",
        "Categories",
        "Genres"
    ]
)

print("Toplam kayıt sayısı:", len(oyun_kontrol))
print("Toplam Playtest sayısı:", len(playtestler))
print("Playtest olmayan kayıt sayısı:", len(normal_oyunlar))
print("Ön kullanılabilir oyun sayısı:", len(kullanilabilir_oyunlar))
print(
    "Eksik alanı bulunan normal kayıt:",
    len(normal_oyunlar) - len(kullanilabilir_oyunlar)
)
```

## Ön kullanılabilir oyun sayısı

Veri setinde toplam 8.139 Playtest kaydı bulunduğu görülmüştür. Playtest kayıtları
çıkarıldığında 117.716 kayıt kalmaktadır. Oyun adı, açıklama, kategori ve tür
alanları eksik olan normal kayıtlar da çıkarıldığında ön kullanılabilir oyun
sayısı 116.281 olmaktadır.

Bu sayı son temiz veri sayısı değildir. Desteklenen dil alanındaki boş değerler
ve tekrar eden oyun kimlikleri ayrıca kontrol edilecektir.

```python
dil_ve_id_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=[
        "AppID",
        "Supported languages"
    ]
)

bos_dil_kayitlari = dil_ve_id_kontrol[
    dil_ve_id_kontrol["Supported languages"] == "[]"
]

tekrar_eden_appid = dil_ve_id_kontrol[
    "AppID"
].duplicated().sum()

print(
    "Desteklenen dil listesi boş olan kayıt:",
    len(bos_dil_kayitlari)
)

print(
    "Tekrar eden AppID sayısı:",
    tekrar_eden_appid
)

print(
    "Benzersiz AppID sayısı:",
    dil_ve_id_kontrol["AppID"].nunique()
)
```

## Dil ve oyun kimliği kontrolü

Desteklenen dil alanında eksik değer görülmemesine rağmen 8.391 kaydın boş liste
olarak tutulduğu belirlenmiştir. Bu kayıtlar dil filtresinde kullanılamayacağı
için veri temizleme aşamasında ayrıca değerlendirilecektir.

AppID alanında tekrar eden bir değere rastlanmamıştır. Veri setindeki 125.855
kaydın tamamı benzersiz bir oyun kimliğine sahiptir.
