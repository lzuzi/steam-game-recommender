## CSV başlık kontrolü

İlk incelemede `Discount` ve `DLC count` sütunlarının başlıkta birleştiği görüldü.
Bu nedenle veri satırları yanlış sütunlarla eşleşiyordu. Birleşmiş başlık iki ayrı
sütuna ayrıldıktan sonra örnek veri 40 sütunla doğru şekilde okundu.

```python
kullanilacak_sutunlar = [
    "AppID",
    "Name",
    "About the game",
    "Price",
    "Windows",
    "Mac",
    "Linux",
    "Supported languages",
    "Categories",
    "Genres"
]

satir_sayisi = 0
eksik_degerler = pd.Series(0, index=kullanilacak_sutunlar)

for parca in pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kullanilacak_sutunlar,
    chunksize=10000
):
    satir_sayisi += len(parca)
    eksik_degerler += parca.isna().sum()

print("Toplam satır sayısı:", satir_sayisi)
eksik_degerler
```

```python
eksik_ozeti = pd.DataFrame({
    "Eksik değer": eksik_degerler,
    "Eksik oranı (%)": (eksik_degerler / satir_sayisi * 100).round(2)
})

eksik_ozeti
```

## İlk değerlendirme

Veri setinde toplam 125.855 oyun kaydı bulunmaktadır. Fiyat, işletim sistemi ve
desteklenen dil alanlarında eksik değer görülmemiştir. Oyun açıklaması, kategori
ve tür alanlarında ise eksik kayıtlar bulunmaktadır. Bu alanlar anlamsal aramada
kullanılacağı için veri temizleme aşamasında ayrıca değerlendirilecektir.

```python
eksik_oyun_listesi = []

for parca in pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kullanilacak_sutunlar,
    chunksize=10000
):
    eksik_mi = parca[
        ["About the game", "Categories", "Genres"]
    ].isna().any(axis=1)

    eksik_kayitlar = parca.loc[
        eksik_mi,
        ["AppID", "Name", "About the game", "Categories", "Genres"]
    ].copy()

    eksik_oyun_listesi.append(eksik_kayitlar)

eksik_oyunlar = pd.concat(eksik_oyun_listesi, ignore_index=True)

print("En az bir alanı eksik oyun sayısı:", len(eksik_oyunlar))
eksik_oyunlar.head(20)
```

```python
playtest_kayitlari = eksik_oyunlar[
    eksik_oyunlar["Name"]
    .fillna("")
    .str.contains("Playtest", case=False)
]

print("Eksik alanı bulunan Playtest sayısı:", len(playtest_kayitlari))
playtest_kayitlari[["AppID", "Name"]].head(20)
```

## Eksik kayıtların incelenmesi

Açıklama, kategori veya tür alanlarından en az biri eksik olan 9.568 kayıt
bulunmaktadır. Bu kayıtların 8.133 tanesinin adında `Playtest` ifadesi yer
almaktadır. Eksik kayıtların büyük bölümünün test sürümlerinden oluştuğu
görülmüştür.

Eksik olan bütün kayıtlar doğrudan silinmeyecektir. Playtest kayıtları ile normal
oyunlar veri temizleme aşamasında ayrı ayrı değerlendirilecektir.

```python
kontrol_sutunlari = [
    "Name",
    "About the game",
    "Categories",
    "Genres"
]

oyun_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=kontrol_sutunlari
)

playtestler = oyun_kontrol[
    oyun_kontrol["Name"].str.contains(
        "Playtest", 
        case=False, 
        na=False 
    )
]

normal_oyunlar = oyun_kontrol[
    oyun_kontrol["Name"].str.contains(
        "Playtest",
        case=False,
        na=False
    ) == False
]

kullanilabilir_oyunlar = normal_oyunlar.dropna(
    subset=[
        "Name",
        "About the game",
        "Categories",
        "Genres"
    ]
)

print("Toplam kayıt sayısı:", len(oyun_kontrol))
print("Toplam Playtest sayısı:", len(playtestler))
print("Playtest olmayan kayıt sayısı:", len(normal_oyunlar))
print("Ön kullanılabilir oyun sayısı:", len(kullanilabilir_oyunlar))
print(
    "Eksik alanı bulunan normal kayıt:",
    len(normal_oyunlar) - len(kullanilabilir_oyunlar)
)
```

## Dil ve oyun kimliği kontrolü

Desteklenen dil alanındaki boş listeler ve AppID alanındaki tekrar eden değerler
kontrol edilmiştir.

```python
dil_ve_id_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=[
        "AppID",
        "Supported languages"
    ]
)

bos_dil_kayitlari = dil_ve_id_kontrol[
    dil_ve_id_kontrol["Supported languages"] == "[]"
]

tekrar_eden_appid = dil_ve_id_kontrol[
    "AppID"
].duplicated().sum()

print(
    "Desteklenen dil listesi boş olan kayıt:",
    len(bos_dil_kayitlari)
)

print(
    "Tekrar eden AppID sayısı:",
    tekrar_eden_appid
)

print(
    "Benzersiz AppID sayısı:",
    dil_ve_id_kontrol["AppID"].nunique()
)
```

Desteklenen dil alanında eksik değer görünmemesine rağmen 8.391 kaydın dil
listesinin boş olduğu belirlenmiştir. Bu kayıtların dili bilinmiyor olarak ele
alınacaktır.

Kullanıcı bir dil tercihi belirtmediğinde bu kayıtlar öneri havuzunda kalabilir.
Belirli bir dil istendiğinde ise doğrulanmış dil eşleşmesi olarak kabul
edilmeyecektir.

AppID alanında tekrar eden bir değere rastlanmamıştır. Veri setindeki 125.855
kaydın tamamı benzersiz bir oyun kimliğine sahiptir.

## Fiyat kontrolü

Fiyat alanında eksik veya negatif değer bulunup bulunmadığı kontrol edilmiştir.
Ücretsiz kayıtların sayısı ve fiyatların genel dağılımı da incelenmiştir.

```python
fiyat_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=[
        "AppID",
        "Name",
        "Price"
    ]
)

print("Fiyatı eksik kayıt sayısı:", fiyat_kontrol["Price"].isna().sum())
print("Negatif fiyatlı kayıt sayısı:", (fiyat_kontrol["Price"] < 0).sum())
print("Ücretsiz kayıt sayısı:", (fiyat_kontrol["Price"] == 0).sum())
print("En düşük fiyat:", fiyat_kontrol["Price"].min())
print("En yüksek fiyat:", fiyat_kontrol["Price"].max())

fiyat_kontrol["Price"].describe()
```

Fiyatı eksik veya negatif olan bir kayda rastlanmamıştır. Veri setinde 26.661
ücretsiz kayıt bulunmaktadır. En yüksek fiyatlı kayıtlar ayrıca kontrol edilmiş
ve fiyatların geçerli olduğu görülmüştür. Bu nedenle fiyat alanında düzeltme
yapılmasına gerek bulunmamaktadır.

## İşletim sistemi kontrolü

Oyunların Windows, Mac ve Linux desteği incelenmiş, hiçbir işletim sistemini
desteklemeyen bir kayıt bulunup bulunmadığı kontrol edilmiştir.

```python
isletim_sistemi_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=[
        "AppID",
        "Name",
        "Windows",
        "Mac",
        "Linux"
    ]
)

isletim_sistemi_yok = isletim_sistemi_kontrol[
    (isletim_sistemi_kontrol["Windows"] == False) &
    (isletim_sistemi_kontrol["Mac"] == False) &
    (isletim_sistemi_kontrol["Linux"] == False)
]

print("Windows destekleyen kayıt:", isletim_sistemi_kontrol["Windows"].sum())
print("Mac destekleyen kayıt:", isletim_sistemi_kontrol["Mac"].sum())
print("Linux destekleyen kayıt:", isletim_sistemi_kontrol["Linux"].sum())
print("İşletim sistemi belirtilmeyen kayıt:", len(isletim_sistemi_yok))

isletim_sistemi_yok.head(20)
```

Bütün kayıtların en az bir işletim sistemini desteklediği görülmüştür. Bu nedenle
işletim sistemi bilgisi nedeniyle çıkarılması gereken bir kayıt bulunmamaktadır.

## Türlerin incelenmesi

Veri setinde bulunan türler ve bu türlerin kullanım sayıları incelenmiştir.
Oyunların yanında yazılım, tasarım, video ve geliştirme araçlarının da veri
setinde bulunduğu görülmüştür.

```python
tur_kontrol = pd.read_csv(
    dosya_yolu,
    skiprows=1,
    names=sutunlar,
    usecols=[
        "AppID",
        "Name",
        "Genres"
    ]
)

turler = (
    tur_kontrol["Genres"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

tur_sayilari = turler.value_counts()

print("Benzersiz tür sayısı:", len(tur_sayilari))

tur_sayilari
```

```python
oyun_disi_turler = [
    "Utilities",
    "Design & Illustration",
    "Animation & Modeling",
    "Video Production",
    "Game Development",
    "Audio Production",
    "Software Training",
    "Photo Editing",
    "Web Publishing",
    "Accounting",
    "Movie",
    "Documentary",
    "Episodic",
    "Short",
    "Tutorial",
    "360 Video"
]

oyun_disi_deseni = "|".join(oyun_disi_turler)

oyun_disi_adaylar = tur_kontrol[
    tur_kontrol["Genres"].str.contains(
        oyun_disi_deseni,
        case=False,
        na=False
    )
]

print("Oyun dışı tür içeren kayıt sayısı:", len(oyun_disi_adaylar))

oyun_disi_adaylar.head(30)
```

```python
temel_oyun_turleri = (
    "Indie|"
    "Casual|"
    "Action|"
    "Adventure|"
    "Simulation|"
    "Strategy|"
    "RPG|"
    "Sports|"
    "Racing|"
    "Massively Multiplayer"
)

guclu_oyun_disi_adaylar = oyun_disi_adaylar[
    oyun_disi_adaylar["Genres"].str.contains(
        temel_oyun_turleri,
        case=False,
        na=False
    ) == False
]

print(
    "Temel oyun türü içermeyen güçlü aday sayısı:",
    len(guclu_oyun_disi_adaylar)
)

guclu_oyun_disi_adaylar.head(30)
```

Oyun dışı türlerden en az birini içeren 1.726 kayıt bulunmuştur. Ancak bazı
kayıtların hem oyun hem de yazılım türü taşıdığı görülmüştür. Bu nedenle temel
oyun türlerinden hiçbirini içermeyen 1.364 kayıt, güçlü oyun dışı aday olarak
ayrılmıştır. Bu kayıtlar temizleme aşamasında değerlendirilecektir.
